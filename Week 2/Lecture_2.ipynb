```python
names = ['Henk', 'Heike', 'Hilda', 'Hans', 'Herman', 'Harish', 'Harshan']
```

```python
print(names)
```

```python
for name in names[2:6]:
    print(name)
```

```python
for name in names[2:6:2]:
    print(name)
```

```python
#Enumerate 
list(enumerate(names))
```

```python
for position, name in enumerate(names):
    print(position, name)
```

```python
#Appending to the list
a = []
for i in range(1,11):
    a.append(i)
```

```python
#List comprehensions
x = [i**2 for i in range(0,11) if i%2 == 0]
print(x)
```

```python
#Nested loops
for i in range(1,11):
    print('Multiplication table for',i)
    for j in range(1,11):
        print(i*j)
    print('')
```
