```python
#Function to calculate the hypotenuse side of a right triangle with sides a and b
def pythagoras(a,b):
    '''Calculates the hypotenuse side of a right triangle with sides a and b
    
    Input arguments: 
    a: (float) a side length
    b: (float) a perpendicular side length

    Returns:
    c: (float) the hypotenuse side length
    '''
    if a>0 and b>0:
        c = (a**2 + b**2)**0.5
        return c
    else:
        print('The side lengths must be larger than 0')
        return None
```

```python
help(pythagoras)
```

```python
pythagoras(3,4)
```

```python
pythagoras(0,4)
```

```python
#Consistency check
pythagoras(1,1) == 2**0.5
```

```python
def countdown(n):
    '''Recursive function that counts down from a positive n to zero'''
    
    #print current value of n
    print(n)
    
    #if you get to zero, you return nothing
    if n == 0:
        return None
    else:
        return countdown(n-1)
```

```python
countdown(3)
```
